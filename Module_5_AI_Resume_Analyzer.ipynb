```python
print("AI Resume Analyzer")
print("===================")
print("Module 5 - AI Tools & Mini Project")
print("Project initialized successfully.")
```

```python
# Resume content to analyze

resume_text = """
Priyanshu Priy
Software Developer | Python Developer | AI/ML Enthusiast

Skills:
Python, SQL, Machine Learning, Pandas, NumPy, Scikit-learn,
Power BI, Git, GitHub, Data Analysis

Projects:
1. Blinkit Data Analysis Dashboard
   Created a Power BI dashboard to analyze sales, outlet performance,
   customer ratings, and product categories.

2. Facial Emotion Recognition
   Developed a machine learning project for recognizing facial emotions
   using image processing and machine learning techniques.

3. Expense Tracker
   Created a Python-based application for tracking personal expenses.

Education:
Bachelor's Degree in Computer Science

Experience:
Completed technical internship projects involving Python,
Machine Learning, Data Analysis, and software testing.
"""

print("Resume loaded successfully.")
print("\nResume Preview:")
print(resume_text)
```

```python
# Skills that the analyzer will look for

skills_database = [
    "Python",
    "SQL",
    "Machine Learning",
    "Pandas",
    "NumPy",
    "Scikit-learn",
    "Power BI",
    "Git",
    "GitHub",
    "Data Analysis",
    "Software Testing",
    "Artificial Intelligence"
]

found_skills = []

for skill in skills_database:
    if skill.lower() in resume_text.lower():
        found_skills.append(skill)

print("Skills Found:")
for skill in found_skills:
    print("✓", skill)

print("\nTotal Skills Found:", len(found_skills))
```

```python
# Simple resume scoring system

score = 0
max_score = 100

# Skills
if len(found_skills) >= 8:
    score += 25
elif len(found_skills) >= 5:
    score += 18
else:
    score += 10

# Projects
project_keywords = ["Projects:", "Project"]
if "Projects:" in resume_text:
    score += 25

# Education
if "Education:" in resume_text:
    score += 20

# Experience
if "Experience:" in resume_text:
    score += 20

# Professional title
if "Developer" in resume_text or "Engineer" in resume_text:
    score += 10

print(f"Resume Score: {score}/{max_score}")
```

```python
# Generate improvement suggestions

suggestions = []

if len(found_skills) < 10:
    suggestions.append(
        "Add more relevant technical skills and tools to strengthen the Skills section."
    )

if "GitHub" not in resume_text:
    suggestions.append(
        "Add your GitHub profile and project repository links."
    )

if "LinkedIn" not in resume_text:
    suggestions.append(
        "Add your LinkedIn profile link."
    )

if "quantified" not in resume_text.lower():
    suggestions.append(
        "Add measurable achievements such as accuracy, percentage improvement, "
        "dataset size, or project impact."
    )

suggestions.append(
    "Use strong action verbs such as developed, implemented, analyzed, automated, and optimized."
)

suggestions.append(
    "Customize the resume keywords according to the specific job description."
)

print("AI Resume Improvement Suggestions")
print("=" * 40)

for i, suggestion in enumerate(suggestions, start=1):
    print(f"{i}. {suggestion}")
```

```python
print("=" * 50)
print("AI RESUME ANALYZER - FINAL ANALYSIS")
print("=" * 50)

print(f"\nResume Score: {score}/100")

print("\nSkills Identified:")
print(", ".join(found_skills))

print("\nImprovement Suggestions:")

for i, suggestion in enumerate(suggestions, start=1):
    print(f"{i}. {suggestion}")

print("\nOverall Feedback:")
print(
    "The resume demonstrates a good foundation in Python, Machine Learning, "
    "Data Analysis, and software development. Adding measurable achievements, "
    "stronger project descriptions, and relevant job-specific keywords can "
    "further improve the resume."
)
```

# Job Description Matching

The application compares the resume with a target job description and identifies relevant skills, missing keywords, and an overall match percentage.

```python
# Target job description

job_description = """
We are looking for a Python Developer with strong programming skills.
The candidate should have experience with Python, SQL, Pandas, NumPy,
Machine Learning, Scikit-learn, Git, GitHub, data analysis and APIs.

Responsibilities include developing Python applications, analyzing data,
building machine learning models, working with databases, writing clean code,
testing applications, and collaborating with software development teams.
"""

print("Job description loaded successfully.")
print("\nJob Description:")
print(job_description)
```

```python
# Extract relevant skills from the job description

job_skills = []

for skill in skills_database:
    if skill.lower() in job_description.lower():
        job_skills.append(skill)

print("Skills Required by Job:")
for skill in job_skills:
    print("✓", skill)

print("\nTotal Required Skills:", len(job_skills))
```

```python
# Compare resume skills with job requirements

matching_skills = []
missing_skills = []

for skill in job_skills:
    if skill.lower() in resume_text.lower():
        matching_skills.append(skill)
    else:
        missing_skills.append(skill)

print("Matching Skills:")
for skill in matching_skills:
    print("✓", skill)

print("\nMissing Skills:")
if missing_skills:
    for skill in missing_skills:
        print("✗", skill)
else:
    print("No major missing skills found.")
```

```python
# Calculate resume-job match percentage

if len(job_skills) > 0:
    match_percentage = (len(matching_skills) / len(job_skills)) * 100
else:
    match_percentage = 0

print(f"Resume-Job Match: {match_percentage:.2f}%")
```

```python
print("=" * 60)
print("AI RESUME ANALYZER - JOB MATCH REPORT")
print("=" * 60)

print(f"\nResume Score: {score}/100")
print(f"Job Match Percentage: {match_percentage:.2f}%")

print("\nMatching Skills:")
if matching_skills:
    print(", ".join(matching_skills))
else:
    print("No matching skills found.")

print("\nMissing Skills:")
if missing_skills:
    print(", ".join(missing_skills))
else:
    print("No major missing skills found.")

print("\nRecommendations:")

if match_percentage >= 80:
    print("✓ Strong match for this job description.")
elif match_percentage >= 60:
    print("✓ Good match, but some relevant keywords should be added.")
else:
    print("✓ Consider improving the resume with more job-relevant skills.")

print("✓ Add measurable achievements to project descriptions.")
print("✓ Use relevant keywords naturally throughout the resume.")
print("✓ Keep project descriptions focused on technologies and results.")
print("✓ Include GitHub and LinkedIn links in the final resume.")
```

```python
# NLP-based Resume and Job Description Similarity

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [resume_text, job_description]

vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(documents)

similarity = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

similarity_percentage = similarity * 100

print("NLP Similarity Analysis")
print("=" * 40)
print(f"Resume-Job Similarity: {similarity_percentage:.2f}%")
```

```python
# Final Resume Analyzer Dashboard

print("=" * 60)
print("              AI RESUME ANALYZER")
print("=" * 60)

print(f"\nResume Score       : {score}/100")
print(f"Job Skill Match    : {match_percentage:.2f}%")
print(f"NLP Similarity     : {similarity_percentage:.2f}%")
print(f"Skills Identified  : {len(found_skills)}")
print(f"Skills Required    : {len(job_skills)}")

print("\n" + "-" * 60)
print("MATCHING SKILLS")
print("-" * 60)

for skill in matching_skills:
    print("✓", skill)

print("\n" + "-" * 60)
print("MISSING SKILLS")
print("-" * 60)

if missing_skills:
    for skill in missing_skills:
        print("✗", skill)
else:
    print("No major missing skills.")

print("\n" + "-" * 60)
print("RECOMMENDATIONS")
print("-" * 60)

print("1. Add measurable achievements to your projects.")
print("2. Include job-specific technical keywords.")
print("3. Keep project descriptions focused on technologies and results.")
print("4. Include GitHub and LinkedIn links.")
print("5. Customize your resume for each target job.")

print("\nAnalysis completed successfully.")
```
